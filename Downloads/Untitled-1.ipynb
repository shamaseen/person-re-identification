```python
import os
for root, dirs, files in os.walk(".", topdown=False):
    
    for file in files:
        if file.split('.')[-1]=='txt':
            file1 = open(os.path.join(root,file), 'r+')
            Lines=file1.readlines()
            try:
                for i in range(len(Lines)):
                    Lines[i]='0'+Lines[i][1:]
            except:
                pass
            file1.close() 
            file1 = open(os.path.join(root,file), 'w+')
            file1.truncate()
            file1.writelines(Lines)
            print(Lines)
            file1.close()
            
```
